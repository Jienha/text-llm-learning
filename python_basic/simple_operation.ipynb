```python

v1 = 10.4

print("\nround up a number:")
print(math.ceil(v1))
print(math.floor(v1))


print("\ndivision in python:")
print(10/2) # float division
print(10//2) # ceil division

print("\nPass a function as an argument in Python")
def add(x, y):
    return x + y

def apply_func(func, a, b):
    return func(a, b)

print(apply_func(add, 3, 5))

print("\nWhat is pass in Python?")
def my_function():
    pass
my_function()


# What is a lambda function? 
# A lambda function is an anonymous function. This function can have any number of parameters but, can have just one statement.
print("\nWhat is a lambda function?")
s1 = "Hello World"
s2 = lambda function: function.upper()
print(s2(s1))

# What is List Comprehension? Give an Example.
print("\nWhat is List Comprehension? Give an Example.")
a = [2,3,4,5]

res = [val ** 2 for val in a]
print(res)

# onther way to do the same thing:
function_bastard = lambda a: a ** 2
print(list(map(function_bastard, a)))

# What are *args and **kwargs
print("\nWhat are *args and **kwargs")
# *args: The special syntax *args in function definitions is used to pass a 
# variable number of arguments to a function. Python program to illustrate
# *args for a variable number of arguments:
def fun(*argv):
    for arg in argv:
        print(arg)
fun('Hello', 'Welcome', 'to', 'GeeksforGeeks')

# **kwargs: The special syntax **kwargs in function definitions is used 
# to pass a variable length argument list. We use the name kwargs with 
# the double star **.
def fun(**kwargs):
    for k, val in kwargs.items():
        print("%s == %s" % (k, val))
fun(s1='Geeks', s2=[1,2,3], s3=True)

# What is a break, continue and pass in Python?
print("\nWhat is a break, continue and pass in Python?")
# break: The break statement terminates the loop containing it.
# continue:  continue statement is opposite to that of the break 
#   statement, instead of terminating the loop, it forces to execute
#   the next iteration of the loop.
# Pass: means performing no operation or in other words, it is a 
#   placeholder in the compound statement, where there should be 
#   a blank left and nothing has to be written there.

# break:
for i in range(10):
    if i == 5:
        break
    print(i)
```

```python
def append_to_list(value, my_list=[]):
    my_list.append(value)
    return my_list

print(append_to_list(1))
print(append_to_list(2))
```

```python
def decorator(func):
    def wrapper():
        print("Before function")
        func()
        print("After function")
    return wrapper

print(decorator('hello'))
```
