## 1. Setup and Imports

```python
# Import necessary libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import PCA
import spacy

# Set up plotting style
plt.style.use('default')
sns.set_palette('husl')

print('✅ All libraries imported successfully!')
```

## 2. TF-IDF Hands-on

TF-IDF (Term Frequency-Inverse Document Frequency) is a classic way to represent text numerically. Let's try it on a sample corpus.

```python
corpus = [
    'Natural language processing is fun and powerful.',
    'I love learning about NLP and machine learning.',
    'Feature extraction is a key step in NLP.',
    'TF-IDF and word embeddings are popular techniques.'
]

vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(corpus)
df = pd.DataFrame(X.toarray(), columns=vectorizer.get_feature_names_out())
df
```

### Visualizing TF-IDF

Let's visualize the TF-IDF scores for the first document.

```python
tfidf_scores = df.iloc[0]
plt.figure(figsize=(10, 4))
sns.barplot(x=tfidf_scores.index, y=tfidf_scores.values)
plt.title('TF-IDF Scores for Document 1')
plt.ylabel('TF-IDF')
plt.xlabel('Term')
plt.show()
```

## 3. Word Embeddings Hands-on

Word embeddings represent words as dense vectors. We'll use spaCy's small English model for this demo.

```python
# Load spaCy's small English model
nlp = spacy.load('en_core_web_sm')

words = ['language', 'processing', 'learning', 'fun', 'powerful']
vectors = np.array([nlp(w).vector for w in words])

print('Shape of embedding for one word:', vectors[0].shape)
vectors
```

### Visualizing Word Embeddings (PCA)

Let's reduce the dimensionality of the embeddings and plot them.

```python
pca = PCA(n_components=2)
reduced = pca.fit_transform(vectors)

plt.figure(figsize=(8, 6))
sns.scatterplot(x=reduced[:,0], y=reduced[:,1])
for i, word in enumerate(words):
    plt.text(reduced[i,0]+0.01, reduced[i,1]+0.01, word)
plt.title('Word Embeddings Visualized (PCA)')
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.show()
```

## 4. Analysis and Comparison

Let's compare the two approaches:
- **TF-IDF**: Good for sparse, interpretable features, but doesn't capture meaning.
- **Embeddings**: Dense, capture semantic similarity, but less interpretable.

Try changing the corpus or the list of words and see how the features change!
